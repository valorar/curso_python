## **Reto 12: La ventana cuadrada del taller**

* Este es un caso de "código roto".
* En el taller de vidrio cortan ventanas cuadradas: el mismo largo en los cuatro lados.
* Crea una función `area_ventana(lado)` que devuelva la superficie de la ventana. La fórmula es `lado ** 2`.
* El siguiente código parece que no funciona correctamente.

```python
def area_ventana(lado)
return lado ** 2
```

#### Ejemplos

1. area_ventana(4) ➞ 16
2. area_ventana(7) ➞ 49
3. area_ventana(0) ➞ 0

````python
# Solución:
def area_ventana(lado)
return lado ** 2
````

````python
print(area_ventana(4))    # 16
print(area_ventana(7))    # 49
print(area_ventana(0))    # 0
````
