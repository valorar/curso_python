## **Reto 83: El panel del depósito de agua**

* Este es un caso de "código roto".
* El panel digital de un depósito de agua muestra el nivel de llenado actual como un texto con porcentaje (por ejemplo, `"60%"`) y guarda también el umbral mínimo de aviso con el mismo formato.
* La función debe devolver `"REPOSTAR"` si el nivel es menor o igual que el umbral y `"SUFICIENTE"` en caso contrario.
* El código actual no funciona bien en algunos casos, por ejemplo al comparar `"9%"` y `"20%"`:

```python
def estado_deposito(nivel, umbral):
    if nivel <= umbral:
        return "REPOSTAR"
    return "SUFICIENTE"
```

#### Ejemplos

1. estado_deposito("25%", "25%") ➞ "REPOSTAR"
2. estado_deposito("35%", "20%") ➞ "SUFICIENTE"
3. estado_deposito("9%", "20%") ➞ "REPOSTAR"
4. estado_deposito("0%", "20%") ➞ "REPOSTAR"

#### Nota

- El símbolo `%` está siempre al final de las cadenas y los números son enteros sin decimales.

````python
# Solución:
def estado_deposito(nivel, umbral):
    if nivel <= umbral:
        return "REPOSTAR"
    return "SUFICIENTE"
````

````python
print(estado_deposito("25%", "25%"))    # REPOSTAR
print(estado_deposito("35%", "20%"))    # SUFICIENTE
print(estado_deposito("9%", "20%"))     # REPOSTAR
print(estado_deposito("0%", "20%"))     # REPOSTAR
````
