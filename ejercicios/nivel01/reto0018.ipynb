## **Reto 18: Conversor de libras a dólares**

* Este es un caso de "código roto".
* Una fintech británica necesita convertir libras esterlinas a dólares estadounidenses.
* El tipo de cambio es 1 libra = 1.2734 dólares.
* El resultado debe redondearse a dos decimales.
* El siguiente código parece que no funciona correctamente.

```python
def libra_a_usd(libras):
    return round(libras * 1,2734)
```

#### Ejemplos

1. libra_a_usd(1) ➞ 1.27
2. libra_a_usd(200) ➞ 254.68
3. libra_a_usd(1_000) ➞ 1273.4

````python
# Solución:
def libra_a_usd(libras):
    return round(libras * 1,2734)
````

````python
print(libra_a_usd(1))        # 1.27
print(libra_a_usd(200))      # 254.68
print(libra_a_usd(1_000))    # 1273.4
````
