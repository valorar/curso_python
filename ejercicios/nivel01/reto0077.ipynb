## **Reto 77: Palabra en frase**

* Este es un caso de "código roto"
* Un buscador de documentos debe detectar si una palabra aparece en un texto.
* Tu compañero usó == en lugar de in y solo encuentra coincidencias exactas.
* Corrige la función. Debe devolver `True` si la palabra está contenida en la frase y `False` en caso contrario.
* No debe distinguir entre mayúsculas y minúsculas.
* Código erróneo:

```python
def contiene(frase, palabra):
    return frase == palabra
```

#### Ejemplos

1. contiene("El gato duerme", "gato") ➞ True
2. contiene("El gato duerme", "perro") ➞ False
3. contiene("Python es genial", "PYTHON") ➞ True

````python
# Solución:
def contiene(frase, palabra):
    return frase == palabra
````

````python
print(contiene("El gato duerme", "gato"))        # True
print(contiene("El gato duerme", "perro"))       # False
print(contiene("Python es genial", "PYTHON"))    # True
````
