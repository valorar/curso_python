## **Reto 39: Turno de limpieza**

* Este es un caso de "código roto".
* En una comunidad de vecinos, los turnos de limpieza se organizan según el número de día.
* En los días múltiplos de 3 se limpia la cocina; en los múltiplos de 5 se limpia el jardín. Cuando coinciden ambos turnos, se realizan las dos tareas. Los demás días se descansa.
* Crea una función `turno_limpieza(dia)` que devuelva `"Cocina"`, `"Jardín"`, `"Cocina y jardín"` o `"Descanso"`, según corresponda.
* El siguiente código no produce siempre el resultado esperado.

```python
def turno_limpieza(dia):
    if dia % 3 == 0:
        return "Cocina"
    elif dia % 5 == 0:
        return "Jardín"
    elif dia % 3 == 0 and dia % 5 == 0:
        return "Cocina y jardín"
    else:
        return "Descanso"
```

#### Ejemplos

1. turno_limpieza(3) ➞ "Cocina"
2. turno_limpieza(10) ➞ "Jardín"
3. turno_limpieza(15) ➞ "Cocina y jardín"
4. turno_limpieza(7) ➞ "Descanso"
5. turno_limpieza(30) ➞ "Cocina y jardín"

#### Nota

- `dia` será un número entero positivo.

````python
# Solución 1: 
def turno_limpieza(dia):
    if dia % 3 == 0:
        return "Cocina"
    elif dia % 5 == 0:
        return "Jardín"
    elif dia % 3 == 0 and dia % 5 == 0:
        return "Cocina y jardín"
    else:
        return "Descanso"
````

````python
print(turno_limpieza(3))     # Cocina
print(turno_limpieza(10))    # Jardín
print(turno_limpieza(15))    # Cocina y jardín
print(turno_limpieza(7))     # Descanso
print(turno_limpieza(30))    # Cocina y jardín
````
