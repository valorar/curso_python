## **Reto 32: Descuento especial**

* Este es un caso de "código roto"
* Una tienda aplica un descuento del 10 % solo a clientes VIP.
* Pero algo salió mal en este código y no funciona bien:
  ```python
  def precio_final(precio, es_vip):
      return precio * 0.90
      if es_vip:
          return precio * 0.90
      return precio
  ```

#### Ejemplos

1. precio_final(100, False) ➞ 100
2. precio_final(100, True) ➞ 90.0
3. precio_final(50, True) ➞ 45.0

````python
# Solución:
def precio_final(precio, es_vip):
    return precio * 0.90
    if es_vip:
        return precio * 0.90
    return precio
````

````python
print(precio_final(100, False))    # 100
print(precio_final(100, True))     # 90.0
print(precio_final(50, True))      # 45.0
````
