## **Reto 21: La taquilla número diez**

* Este es un caso de "código roto".
* En el gimnasio hay una taquilla antigua que solo se abre si el número tecleado es exactamente el 10.
* El código que tenemos no funciona:

```python
def es_taquilla_diez(numero):
    return numero = 10
```

* La función debería devolver `True` solo cuando el número sea 10.
* Cualquier otro número, incluido el 0, debería devolver `False`.

#### Ejemplos

1. es_taquilla_diez(3) ➞ False
2. es_taquilla_diez(10) ➞ True
3. es_taquilla_diez(0) ➞ False
4. es_taquilla_diez(25) ➞ False

````python
# Solución:
def es_taquilla_diez(numero):
    return numero = 10
````

````python
print(es_taquilla_diez(3))     # False
print(es_taquilla_diez(10))    # True
print(es_taquilla_diez(0))     # False
print(es_taquilla_diez(25))    # False
````
