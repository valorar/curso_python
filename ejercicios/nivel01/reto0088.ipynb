## **Reto 88: Código sin inicial**

* Este es un caso de "código roto".
* Una app de juegos genera códigos de acceso eliminando la primera letra del nombre del jugador.
* El siguiente código tiene un error y no produce el resultado esperado.

```python
def codigo_acceso(nombre):
    return nombre[0:]
```

#### Ejemplos

1. codigo_acceso("Mario") ➞ "ario"
2. codigo_acceso("Elena") ➞ "lena"
3. codigo_acceso("Bruno") ➞ "runo"

````python
# Solución:
def codigo_acceso(nombre):
    return nombre[0:]
````

````python
print(codigo_acceso("Mario"))    # ario
print(codigo_acceso("Elena"))    # lena
print(codigo_acceso("Bruno"))    # runo
````
