## **Reto 42: Vida del personaje**

* Este es un caso de "código roto"
* Estás programando un juego RPG (role-playing game) y necesitas una función que compruebe si un personaje sigue vivo.
* Si su vida es mayor que 0, debe devolver `True`; en caso contrario, `False`.
* Pero algo falla en este código:

```python
def esta_vivo(vida):
    if vida > 0:
        return False
```

#### Ejemplos

1. esta_vivo(10) ➞ True
2. esta_vivo(0) ➞ False
3. esta_vivo(-5) ➞ False

````python
# Solución 1: 
def esta_vivo(vida):
    if vida > 0:
        return False
````

````python
print(esta_vivo(10))    # True
print(esta_vivo(0))     # False
print(esta_vivo(-5))    # False
````
