## **Reto 64: Crecimiento bacteriano**

* Este es un caso de "código roto"
* Una colonia de bacterias se duplica cada hora. Tu compañero ha escrito una fórmula que no representa esa duplicación.
* Corrige la función. Debe recibir el número de horas y devolver la cantidad de bacterias, asumiendo que se parte de 1.
* Código erróneo:

```python
def bacterias(horas):
    return 1 + horas * 2
```

#### Ejemplos

1. bacterias(0) ➞ 1
2. bacterias(1) ➞ 2
3. bacterias(3) ➞ 8
4. bacterias(24) ➞ 16777216

#### Nota

- El número de horas es un entero mayor o igual que cero.

````python
# Solución:
def bacterias(horas):
    return 1 + horas * 2
````

````python
print(bacterias(0))     # 1
print(bacterias(1))     # 2
print(bacterias(3))     # 8
print(bacterias(24))    # 16777216
````
