## **Reto 98: Tarifa de aparcamiento**

* Este es un caso de "código roto".
* Una app de aparcamiento calcula la tarifa según el tipo de vehículo: "moto" paga 1.5 € por hora, "coche" paga 2.5 € por hora, y cualquier otro vehículo devuelve `None`.
* El siguiente código tiene errores y no funciona correctamente.

```python
def tarifa(horas, vehiculo):
    if vehiculo.lower() = "moto":
        return horas * 1.5
    elif vehiculo.upper() = "coche":
        return horas * 2.5
    else:
        return none
```

#### Ejemplos

1. tarifa(2, "moto") ➞ 3.0
2. tarifa(3, "coche") ➞ 7.5
3. tarifa(1, "Moto") ➞ 1.5
4. tarifa(4, "camión") ➞ None

````python
# Solución:
def tarifa(horas, vehiculo):
    if vehiculo.lower() = "moto":
        return horas * 1.5
    elif vehiculo.upper() = "coche":
        return horas * 2.5
    else:
        return none
````

````python
print(tarifa(2, "moto"))      # 3.0
print(tarifa(3, "coche"))     # 7.5
print(tarifa(1, "Moto"))      # 1.5
print(tarifa(4, "camión"))    # None
````
