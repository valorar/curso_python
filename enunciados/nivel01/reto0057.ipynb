## **Reto 57: Rutas de reparto duplicadas**

* Este es un caso de "código roto".
* Una empresa de logística guarda cada ruta como una lista ordenada de paradas.
* Dos rutas están duplicadas únicamente si contienen las mismas paradas en el mismo orden. Recorrer los mismos lugares en otro orden supone realizar una ruta diferente.
* Crea una función `misma_ruta(ruta1, ruta2)` que devuelva `True` cuando las dos rutas sean iguales y `False` cuando no.
* El siguiente código tiene errores y no produce los resultados esperados.

```python
def misma_ruta(ruta1, ruta2):
    if [ruta1 = ruta2]:
        print(true)
    else:
        print(false)
```

#### Ejemplos

1. misma_ruta(["almacén", "centro"], ["almacén", "centro"]) ➞ True
2. misma_ruta(["almacén", "centro"], ["centro", "almacén"]) ➞ False
3. misma_ruta(["almacén", "centro"], ["almacén", "aeropuerto"]) ➞ False
4. misma_ruta(["almacén", "centro"], ["almacén", "centro", "estación"]) ➞ False

````python
# Solución:
def misma_ruta(ruta1, ruta2):
    if [ruta1 = ruta2]:
        print(true)
    else:
        print(false)
````

````python
print(misma_ruta(["almacén", "centro"], ["almacén", "centro"]))                # True
print(misma_ruta(["almacén", "centro"], ["centro", "almacén"]))                # False
print(misma_ruta(["almacén", "centro"], ["almacén", "aeropuerto"]))            # False
print(misma_ruta(["almacén", "centro"], ["almacén", "centro", "estación"]))    # False
````
