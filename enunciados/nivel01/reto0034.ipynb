## **Reto 34: Alarma de seguridad**

* Este es un caso de "código roto".
* Un sistema de seguridad usa tres puertas lógicas (AND, OR y NOT) para controlar el acceso. Trabajan con los enteros 1 y 0, y deben devolver siempre 1 o 0.
* El código tiene errores y no produce los resultados esperados.

```python
def puerta_and(a, b):
    return a or b

def puerta_or(a, b):
    return a and b

def puerta_not(a):
    return not a
```

#### Ejemplos

1. puerta_and(0, 1) ➞ 0
2. puerta_and(0, 0) ➞ 0
3. puerta_or(1, 1) ➞ 1
4. puerta_or(0, 1) ➞ 1
5. puerta_not(1) ➞ 0
6. puerta_not(0) ➞ 1


````python
# Solución:
def puerta_and(a, b):
    return a or b

def puerta_or(a, b):
    return a and b

def puerta_not(a):
    return not a
````

````python
print(puerta_and(0, 1))    # 0
print(puerta_and(0, 0))    # 0
print(puerta_or(1, 1))     # 1
print(puerta_or(0, 1))     # 1
print(puerta_not(1))       # 0
print(puerta_not(0))       # 1
````
