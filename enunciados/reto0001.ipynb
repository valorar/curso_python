## **Reto 1: Suma de puntos**

* En un torneo, un equipo consigue puntos en dos rondas.
* Escribe una función `suma_puntos(a, b)` que reciba los puntos obtenidos en cada ronda y devuelva la puntuación total del equipo.

#### Ejemplos

1. suma_puntos(9, 4) ➞ 13
2. suma_puntos(3, 8) ➞ 11
3. suma_puntos(10, 0) ➞ 10
4. suma_puntos(-3, 1) ➞ -2
